```python
import requests


def consultar_cep():
    cep = input("Digite o CEP somente com números: ")

    if len(cep) != 8 or not cep.isdigit():
        print("CEP inválido.")
        return

    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        resposta = requests.get(url, timeout=10)

        if resposta.status_code != 200:
            print("Erro ao consultar o CEP.")
            return

        dados = resposta.json()

        if "erro" in dados:
            print("CEP não encontrado.")
            return

        print("\n=== ENDEREÇO ENCONTRADO ===")
        print(f"CEP: {dados.get('cep')}")
        print(f"Logradouro: {dados.get('logradouro')}")
        print(f"Bairro: {dados.get('bairro')}")
        print(f"Cidade: {dados.get('localidade')}")
        print(f"Estado: {dados.get('uf')}")

    except requests.RequestException:
        print("Não foi possível acessar o serviço de consulta.")


while True:

    print("\n==============================")
    print("      CONSULTA DE CEP")
    print("==============================")

    print("1 - Consultar CEP")
    print("2 - Sair")

    opcao = input("\nEscolha uma opção: ")

    if opcao == "1":
        consultar_cep()

    elif opcao == "2":
        print("\nPrograma encerrado.")
        break

    else:
        print("\nOpção inválida.")
```
